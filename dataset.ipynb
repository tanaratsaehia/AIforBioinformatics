```python
import pandas as pd

# 1. โหลดไฟล์ CSV (เปลี่ยน 'map_file.csv' เป็นชื่อไฟล์จริงของคุณ)
# header=None คือไม่มีหัวตาราง, names คือเราตั้งชื่อคอลัมน์เองให้เรียกง่าย
df_map = pd.read_csv(r'C:\Users\User\Documents\GitHub\AIforBioinformatics\ignore_dir\dataset\map_probe.csv', header=None, names=['probe', 'gene'])

# 2. แปลงเป็น Dictionary { '1007_s_at': 'MIR4640', ... }
mapping_dict = dict(zip(df_map['probe'], df_map['gene']))
```

```python
def map_selected_genes(text):
    if pd.isna(text):
        return text
    
    # แยก ID ออกจากกัน (และตัดช่องว่างทิ้ง)
    ids = [i.strip() for i in text.split(',')]
    
    # เอาแต่ละ ID ไปเทียบใน dict ถ้าไม่เจอให้ใช้ ID เดิม (ป้องกันข้อมูลหาย)
    translated = [mapping_dict.get(i, i) for i in ids]
    
    # รวมกลับเป็น string เดียวกัน คั่นด้วยคอมม่าเหมือนเดิม
    return ", ".join(translated)

# นำไปใช้งานกับ DataFrame หลักของคุณ (สมมติชื่อ df)
sum_df['Selected_Genes'] = sum_df['Selected_Genes'].apply(map_selected_genes)
```
